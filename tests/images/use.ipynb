```python
import requests
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np
import os
from pathlib import Path


def query_similar_images(image_path, host="http://127.0.0.1", port=5002, top_k=3):
    """
    向 Flask 服务的 /find_similar 接口发送图像，查找最相似的图像。

    参数:
        image_path (str): 本地图像文件路径（如 'test_query.jpg'）
        host (str): 服务器主机地址
        port (int): 服务器端口
        top_k (int): 返回最相似的前 k 个结果

    返回:
        dict: 包含 results 或 error 信息的响应
    """
    url = f"{host}:{port}/find_similar"

    payload = {
        "input_image": image_path,
        "top_k": top_k
    }

    try:
        response = requests.post(url, json=payload, timeout=10)
        if response.status_code == 200:
            return response.json()
        else:
            return {"error": f"HTTP {response.status_code}: {response.text}"}
    except requests.exceptions.RequestException as e:
        return {"error": f"请求失败: {str(e)}"}


def visualize_similar_images(query_image_path, similar_results, top_k=3):
    """
    可视化查询图像和相似图像结果。

    参数:
        query_image_path (str): 查询图像的本地路径
        similar_results (list): 来自 /find_similar 接口的结果，格式如:
                                [(spot_id, image_path, similarity), ...]
        top_k (int): 显示前 k 个相似结果
    """
    if not os.path.exists(query_image_path):
        print(f"❌ 查询图像不存在: {query_image_path}")
        return

    if len(similar_results) == 0:
        print("⚠️ 未找到相似图像")
        return

    results = similar_results[:top_k]

    n_cols = top_k + 1
    fig, axes = plt.subplots(1, n_cols, figsize=(5 * n_cols, 5))
    if n_cols == 1:
        axes = [axes]  # 兼容单子图情况

    query_img = mpimg.imread(query_image_path)
    axes[0].imshow(query_img)
    axes[0].axis('off')

    for idx, item in enumerate(results):
        # item 格式: (spot_id, image_path, similarity)
        spot_id, image_path, similarity = item

        image_path = str(image_path)

        if not os.path.exists(image_path):
            axes[idx + 1].imshow(np.zeros((100, 100, 3)))  
        else:
            result_img = mpimg.imread(image_path)
            axes[idx + 1].imshow(result_img)
        axes[idx + 1].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
query_image_path = "/home/zhangbi/Zhangbi_Traveler/Figure_to_text/images/didaopoumiantu_12.jpeg"

result = query_similar_images(query_image_path, top_k=10)

if "error" in result:
    print(f"❌ 错误: {result['error']}")
else:
    print("✅ 找到相似图像:")
    for i, item in enumerate(result["results"], 1):
        spot_id, img_path, sim = item
        print(f"  {i}. [{spot_id}] {os.path.basename(img_path)} | 相似度: {sim:.3f}")

    visualize_similar_images(
        query_image_path=query_image_path,
        similar_results=result["results"],
        top_k=10
    ) 
```

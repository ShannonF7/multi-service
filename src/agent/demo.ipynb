```python
import json


def parse_graph_detail(detail):

    context = {
        "entity": {},
        "properties": [],
        "relations": [],
        "neighbors": []
    }


    # 实体
    root = detail.get("root")

    if root:
        context["entity"] = {
            "id": root.get("node_id"),
            "name": root.get("name"),
            "type": root.get("node_type"),
            "description": root.get("description"),
        }


    # 属性
    for fact in detail.get("facts", []):

        item = {
            "name": fact.get("key"),
            "value": fact.get("value"),
        }


        provenance = fact.get(
            "provenance_json"
        )

        if provenance:
            try:
                p = json.loads(provenance)

                if p:
                    item["evidence"] = {
                        "quote": p[0].get("quote"),
                        "source": p[0].get("source_title"),
                        "confidence": p[0].get("confidence")
                    }

            except:
                pass


        context["properties"].append(item)


    # 关系
    for r in detail.get("relations", []):

        context["relations"].append({
            "type": r.get("relation_label"),
            "target": r.get("target_name")
        })


    # 邻居
    for n in detail.get("neighbors", []):

        context["neighbors"].append({
            "name": n.get("name"),
            "id": n.get("node_id")
        })


    return context
```

```python
import json


def parse_fact(fact):

    result = {
        "property": fact.get("key"),
        "value": fact.get("value"),
        "evidence": None
    }


    raw = fact.get("provenance_json")

    if raw:
        try:
            evidence_list = json.loads(raw)

            if evidence_list:

                e = evidence_list[0]

                result["evidence"] = {
                    "quote": e.get("quote"),
                    "source": e.get("source_title"),
                    "confidence": e.get("confidence"),
                    "source_type": e.get("source_type")
                }

        except Exception as ex:
            print(ex)


    return result
```

```python
def build_graph_context(detail):

    return {

        "entity":{
            "id":detail["node_id"],
            "name":detail["root"]["name"],
        },


        "facts":[
            parse_fact(f)
            for f in detail["facts"]
        ],


        "relations":
            detail["relations"],


        "neighbors":
            detail["neighbors"]
    }
```
